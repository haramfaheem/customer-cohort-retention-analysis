```python
import pandas as pd
import os
from google.colab import drive
drive.mount('/content/drive')

p = '/content/drive/MyDrive/archive (2)'
print(os.listdir(p))
```

```python
orders    = pd.read_csv(p+'olist_orders_dataset.csv')
items     = pd.read_csv(p+'olist_order_items_dataset.csv')
payments  = pd.read_csv(p+'olist_order_payments_dataset.csv')
reviews   = pd.read_csv(p+'olist_order_reviews_dataset.csv')
customers = pd.read_csv(p+'olist_customers_dataset.csv')
products  = pd.read_csv(p+'olist_products_dataset.csv')
trans     = pd.read_csv(p+'product_category_name_translation.csv')

for name, d in [('orders',orders),('items',items),('payments',payments),
                ('reviews',reviews),('customers',customers),('products',products),('trans',trans)]:
    print(name, d.shape)
```

```python
pay_agg = payments.groupby('order_id').agg(
    payment_value=('payment_value', 'sum'),
    payment_installments=('payment_installments', 'max'),
    payment_type=('payment_type', 'first')
).reset_index()

rev_agg = reviews.groupby('order_id').agg(
    review_score=('review_score', 'mean'),
    review_comment_message=('review_comment_message', 'first')
).reset_index()

print(pay_agg.shape, rev_agg.shape)
print(pay_agg['order_id'].is_unique, rev_agg['order_id'].is_unique)
```

```python
products = products.merge(trans, on='product_category_name', how='left')
print(products.shape)
```

```python
df = (orders
      .merge(customers, on='customer_id', how='left')
      .merge(items, on='order_id', how='left')
      .merge(products, on='product_id', how='left')
      .merge(pay_agg, on='order_id', how='left')
      .merge(rev_agg, on='order_id', how='left'))

print(df.shape)
```

```python
print("Total rows:", len(df))
print("Unique orders:", df['order_id'].nunique())
print("Duplicate (order_id + item_id):",
      df.duplicated(subset=['order_id','order_item_id']).sum())
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
df.to_csv(p+'olist_merged.csv', index=False)
```

```python
date_cols = ['order_purchase_timestamp', 'order_approved_at',
             'order_delivered_carrier_date', 'order_delivered_customer_date',
             'order_estimated_delivery_date', 'shipping_limit_date']
for c in date_cols:
    df[c] = pd.to_datetime(df[c])

print(df[date_cols].dtypes)
```

```python

df = df.dropna(subset=['product_id', 'price']).copy()


df['review_comment_message'] = df['review_comment_message'].fillna('No comment')
df['product_category_name_english'] = (df['product_category_name_english']
                                       .fillna(df['product_category_name'])
                                       .fillna('unknown'))
df['product_category_name'] = df['product_category_name'].fillna('unknown')


for c in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
    df[c] = df[c].fillna(df[c].median())


for c in ['product_name_lenght', 'product_description_lenght', 'product_photos_qty']:
    df[c] = df[c].fillna(df[c].median())
```

```python
print(df.duplicated(subset=['order_id', 'order_item_id']).sum())
```

```python
delivered = df[df['order_status'] == 'delivered'].dropna(subset=['order_delivered_customer_date']).copy()
print(delivered.shape)
```

```python
delivered['delivery_days'] = (delivered['order_delivered_customer_date'] - delivered['order_purchase_timestamp']).dt.days
delivered['delay_days']    = (delivered['order_delivered_customer_date'] - delivered['order_estimated_delivery_date']).dt.days
delivered['is_late']       = (delivered['delay_days'] > 0).astype(int)
delivered['order_month']   = delivered['order_purchase_timestamp'].dt.to_period('M')
delivered['total_value']   = delivered['price'] + delivered['freight_value']
```

```python
print(delivered.shape)
print(delivered[['delivery_days', 'delay_days', 'is_late', 'total_value']].describe())
missing = delivered.isna().sum()
print(missing[missing > 0])
```

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from scipy import stats

order_df = delivered.groupby('order_id').agg(
    review_score=('review_score', 'first'),
    delivery_days=('delivery_days', 'first'),
    delay_days=('delay_days', 'first'),
    is_late=('is_late', 'first'),
    price=('price', 'sum'),
    freight_value=('freight_value', 'sum'),
    n_items=('order_item_id', 'count'),
    weight_g=('product_weight_g', 'sum'),
    payment_installments=('payment_installments', 'first'),
    payment_type=('payment_type', 'first'),
    customer_state=('customer_state', 'first'),
).reset_index().dropna(subset=['review_score'])

order_df['low_review'] = (order_df['review_score'] <= 2).astype(int)
print(order_df.shape)
order_df.describe()
```

```python
num_cols = ['delivery_days','delay_days','price','freight_value','weight_g','payment_installments','n_items']

print(order_df[num_cols].agg(['skew','kurtosis']).T)
for c in ['delivery_days','price','freight_value']:
    fig, ax = plt.subplots(1, 2, figsize=(11,3.5))
    sns.histplot(order_df[c], kde=True, ax=ax[0]); ax[0].set_title(f'{c} histogram')
    stats.probplot(order_df[c], dist='norm', plot=ax[1]); ax[1].set_title(f'{c} QQ plot')
    plt.show()
for c in ['delivery_days','price','freight_value']:
    s = order_df[c].sample(5000, random_state=42)
    print(c, '| Shapiro p =', stats.shapiro(s).pvalue,
             '| D\'Agostino p =', stats.normaltest(order_df[c]).pvalue)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler


skewed = ['price','freight_value','weight_g','delivery_days']
for c in skewed:
    order_df[c+'_log'] = np.log1p(order_df[c])

print(order_df[[c+'_log' for c in skewed]].skew())


log_cols = [c+'_log' for c in skewed]
std_scaled = pd.DataFrame(StandardScaler().fit_transform(order_df[log_cols]), columns=log_cols)
mm_scaled  = pd.DataFrame(MinMaxScaler().fit_transform(order_df[log_cols]), columns=log_cols)
print(std_scaled.describe().loc[['mean','std']])
```

```python

plt.figure(figsize=(8,6))
sns.heatmap(order_df[num_cols + ['review_score']].corr(method='spearman'),
            annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Spearman correlation')
plt.show()


from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

X = sm.add_constant(order_df[num_cols].dropna())
vif = pd.DataFrame({'feature': X.columns,
                    'VIF': [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]})
print(vif[vif['feature'] != 'const'].sort_values('VIF', ascending=False))
```

```python

for c in num_cols:
    r, p = stats.spearmanr(order_df[c], order_df['review_score'])
    print(f'{c:22s} rho = {r:6.3f}   p = {p:.4g}')


ct = pd.crosstab(order_df['is_late'], order_df['low_review'])
print(ct)
chi2, p, dof, exp = stats.chi2_contingency(ct)
n = ct.values.sum()
cramers_v = np.sqrt(chi2 / (n * (min(ct.shape) - 1)))
print(f'Chi2 p-value = {p:.4g}, Cramer\'s V = {cramers_v:.3f}')
```

```python
on_time = order_df.loc[order_df['is_late']==0, 'review_score']
late    = order_df.loc[order_df['is_late']==1, 'review_score']

print('Std on-time:', on_time.std(), '| Std late:', late.std())
print('Levene test:', stats.levene(on_time, late))
```

```python
def cohens_d(a, b):
    na, nb = len(a), len(b)
    sp = np.sqrt(((na-1)*a.var(ddof=1) + (nb-1)*b.var(ddof=1)) / (na+nb-2))
    return (a.mean() - b.mean()) / sp


print('Mean on-time:', on_time.mean(), '| Mean late:', late.mean())
print('Welch t-test      :', stats.ttest_ind(on_time, late, equal_var=False))
print('Mann-Whitney U    :', stats.mannwhitneyu(on_time, late))
print('Cohen\'s d         :', cohens_d(on_time, late))


groups = [g['review_score'].values for k, g in order_df.groupby('payment_type') if k != 'not_defined']
print('ANOVA   :', stats.f_oneway(*groups))
print('Kruskal :', stats.kruskal(*groups))
print(order_df.groupby('payment_type')['review_score'].agg(['mean','count']))
```

```python
sk_raw = order_df[['price','freight_value','weight_g','delivery_days']].skew().abs().max()
sk_log = order_df[['price_log','freight_value_log','weight_g_log','delivery_days_log']].skew().abs().max()
X = sm.add_constant(order_df[num_cols].dropna())
max_vif = max(variance_inflation_factor(X.values, i) for i in range(1, X.shape[1]))
lev_p = stats.levene(on_time, late).pvalue
d = cohens_d(on_time, late)

checks = pd.DataFrame([
    ['Unique orders = 99441',      df['order_id'].nunique(),            df['order_id'].nunique() == 99441],
    ['No duplicate item rows',     df.duplicated(['order_id','order_item_id']).sum(), df.duplicated(['order_id','order_item_id']).sum() == 0],
    ['Raw skew < 1',               round(sk_raw,2),                     sk_raw < 1],
    ['Log skew < 1',               round(sk_log,2),                     sk_log < 1],
    ['Max VIF < 5',                round(max_vif,2),                    max_vif < 5],
    ['Equal variance (Levene>0.05)', f'{lev_p:.2g}',                    lev_p > 0.05],
    ['Effect size d > 0.8',        round(d,2),                          abs(d) > 0.8],
], columns=['Check','Value','Pass'])
print(checks)
```
